## 6. *_like()

```python
x = torch.empty(2, 2, 3)
print(x.shape)
print(x)

empty_like_x = torch.empty_like(x)
print(empty_like_x.shape)
print(empty_like_x)

zeros_like_x = torch.zeros_like(x)
print(zeros_like_x.shape)
print(zeros_like_x)

ones_like_x = torch.ones_like(x)
print(ones_like_x.shape)
print(ones_like_x)

rand_like_x = torch.rand_like(x)
print(rand_like_x.shape)
print(rand_like_x)
```

## 14. Concatenating tensors 

```python
# torch.cat
x = torch.tensor([[1, 2], [3, 4]])
y = torch.tensor([[5, 6]])

# Конкатенація вздовж 0-го виміру
result = torch.cat((x, y), dim=0)
print(result)
```

```python
# torch.cat
x = torch.tensor([[1, 2], [3, 4]])
y = torch.tensor([[5, 6], [7, 8]])

# Конкатенація вздовж 0-го виміру
result = torch.cat((x, y), dim=0)
print(result)
```

```python
# torch.cat
x = torch.tensor([[1, 2], [3, 4]])
y = torch.tensor([[5, 6], [7, 8]])

# Конкатенація вздовж 1-го виміру
result = torch.cat((x, y), dim=1)
print(result)
```

## 15. Stack tensors 

```python
#torch.stack
x = torch.tensor([[1, 2], [3, 4]])
y = torch.tensor([[5, 6], [7, 8]])

# Стек вздовж нового виміру (вимір 0)
result = torch.stack((x, y), dim=0)
print(result)
```

```python
#torch.stack
x = torch.tensor([[1, 2], [3, 4]])
y = torch.tensor([[5, 6], [7, 8]])

# Стек вздовж нового виміру (вимір 1)
result = torch.stack((x, y), dim=1)
print(result)
```

## 22. PyTorch and NumPy

## 24. Working with tensor dimensions ( squeeze() and unsqueeze() )

```python
a = torch.rand(3, 226, 226)
b = a.unsqueeze(0)

print(a.shape)
print(b.shape)
```

```python
a = torch.rand(1, 20)
print(a.shape)
print(a)

b = a.squeeze(0)
print(b.shape)
print(b)

c = torch.rand(2, 2)
print(c.shape)

d = c.squeeze(0)
print(d.shape)
```

```python
a = torch.ones(4, 3, 2)

c = a * torch.rand(3, 1)
print(c)
```

```python
a = torch.ones(4, 3, 2)
b = torch.rand(3)
print(a)
print(b)
```

```python
a = torch.ones(4, 3, 2)
b = torch.rand(3)       # спроба помножити a * b призведе до помилки під час виконання
c = b.unsqueeze(1)      # змінити на двовимірний тензор, додавши в кінці новий dim (a.shape = [4,3,2], c.shape = [3,1])

print(c.shape)
print(a * c)            # трансляція знову працює!
```

## 25. Reshaping tensors

```python
output3d = torch.rand(6, 20, 20)
print(output3d.shape)

input1d = output3d.reshape(6 * 20 * 20)
print(input1d.shape)

# також може викликати його як метод у модулі torch:
print(torch.reshape(output3d, (6 * 20 * 20,)).shape)
```

```python
# OPTIONAL — added for understanding reshape/view

x = torch.tensor([[1, 2], [3, 4]])

y = x.reshape(4)

print(x)
print(y)

y[0] = 100

print(x)
print(y)
```

```python
# OPTIONAL — added for understanding clone()

x = torch.tensor([[1, 2], [3, 4]])

y = x.reshape(4).clone()

y[0] = 100

print(x)
print(y)
```
